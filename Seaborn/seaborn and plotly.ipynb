```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
student = pd.read_csv('sns_data (1).csv')
student.head()
```

```python
#axes Level
sns.lineplot(data=student, x='week', y='attendance_rate',hue='gender')
```

```python
sns.lineplot(data=student, x='week', y='attendance_rate',hue='gender',errorbar=None)
```

```python
#figure level kind =like ditea hoi karon dot level asa

sns.relplot(kind='line', data=student, x='week', y='attendance_rate',hue='subject',errorbar=None)  #hue catagory change korea
```

### Scatter Plot

```python
sns.scatterplot(data=student, x='week', y='attendance_rate',hue='subject',style='subject')
```

```python
sns.scatterplot(data=student, x='attendance_rate',y='test_score',hue='gender',size='Tshirt_size')
```

```python
# Load the tips dataset
tips = sns.load_dataset("tips")

print(tips.head())
```

```python
sns.scatterplot(data=tips, x='total_bill',y='tip',style='sex',size='size')
```

### Facet Plot

```python
sns.relplot(kind='scatter',data=student, x='study_hours', y='test_score',col='week')
```

```python
sns.relplot(kind='scatter',data=student, x='study_hours', y='test_score',col='gender')
```

### Histogram 

```python
# Histogram  work on Neumerical valuea
#for axis

sns.histplot(data = student, x = 'attendance_rate', hue = 'gender')
```

```python
#for figure

sns.displot(kind='hist',data = student, x = 'study_hours',col='gender',color='red',edgecolor='black',bins=13)
```

### KDE plot

```python
# karnel density estimation plot

sns.kdeplot(data = student, x = 'attendance_rate')
```

```python
sns.kdeplot(data = tips, x = 'tip',hue='sex',fill=True)
```

```python
sns.histplot(data = tips, x = 'tip')
```

### Pairplot and Jointplot 

```python
#pair Plot

df = pd.read_csv('student_dataset_complete.csv')
df
```

```python
student_marks = df[['Marks','study_hours','attendance_rate','gender']]
student_marks
```

```python
sns.pairplot(data=student_marks,hue='gender')
```

```python
#joint  plot

sns.jointplot(data=student_marks, x= 'study_hours', y= 'Marks',hue='gender',kind='kde')
```
