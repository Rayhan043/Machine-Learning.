### NDarray

```python
import numpy as np

#1D array

arr1 = np.array([1,2,3,4,5,])
print(arr1)
print(type(arr1))

print(arr1.ndim) #koto dimentional a aca ta dekhar jonno



#2d array
arr2 = np.array([ [1,2,3],
                  [4,5,6] 
                ])
print(arr2.ndim)


# 3D array

arr3 = np.array([ 
            [     [1,2,3],#floor 1
                  [4,5,6] ],

            [     [1,2,3],#Floor 2
                  [4,5,6] ],
            [     [1,2,3],#Floor 3
                  [4,5,6] ]
                ])

print(arr3.ndim)
print(arr3)
```

### Array attrubutes

```python
#Dimention Check
print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
print("\n")

#shape
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
print("\n")

#Data Type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)
print("\n")

#size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

```python
#Dictionary
dc = {'a' : 10, 'b' : 20, 'c' : 30}
keys = dc.keys()
print(keys)
values = dc.values()
print(values)

items = dc.items()
print(items)
```

```python

#keys to nd array
arr = np.array(list(keys))
print(arr)
# arr = arr.astype(np.int16)   
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#Values To nd Array
arr = np.array(list(values))
print(arr)
arr = arr.astype(np.int16)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#items to nd array

arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```
