### 2D  line poting using Data

```python
df = pd.read_csv('enrollment_data.csv')
df
```

### Scatter Plots 

```python
#very important for Ml algorithm
student = pd.read_csv('student_IQdata.csv')
student
```

```python
#Relation With IQ score and study hours is positive correlation 
plt.xlabel('Study Hour')
plt.ylabel('IQ Score')
plt.title('Relation With IQ score and study hours')
plt.scatter(student['Study_Hour'], student['IQ_Score'],color='black',label='Study and IQ')
```

```python
#Relation With IQ score and shoe size is positive correlation 
plt.xlabel('shoe size')
plt.ylabel('IQ Score')
plt.title('Relation With IQ score and shoe size')
plt.scatter(student['Shoe_Size'], student['IQ_Score'],color='red')
```

```python
#Relation With IQ score and Chilling hours is 
plt.xlabel('Chilling Hours')
plt.ylabel('IQ Score')
plt.title('Relation With IQ score and Chilling Hours')
plt.scatter(student['Chilling_Hours'], student['IQ_Score'],color='maroon')
```

```python
# both 3 together

#Relation With IQ score and study hours is positive correlation 
plt.xlabel('Study Hour')
plt.ylabel('IQ Score')
plt.title('Relation With IQ score and Both three')
plt.scatter(student['Study_Hour'], student['IQ_Score'],color='black',label='Study and IQ')


#Relation With IQ score and shoe size is No correlation 
plt.xlabel('shoe size')
plt.ylabel('IQ Score')
plt.scatter(student['Shoe_Size'], student['IQ_Score'],color='red',label='shoe size and IQ')


#Relation With IQ score and shoe size is Negetive correlation 
plt.xlabel('Chilling Hours')
plt.ylabel('IQ Score')
plt.scatter(student['Chilling_Hours'], student['IQ_Score'],color='blue',label='chilling hours and IQ')
plt.legend()
```

### Bar Chart

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
df
```

```python
gender = df.groupby('Gender').size()
gender.index
gender.values
```

```python
plt.bar(gender.index, gender.values,color=['pink','black'],edgecolor='red')
plt.xlabel("Gender")
plt.ylabel("count")
plt.title('Count of Male and Female Bar')
```

```python
df = pd.read_csv('student_data.csv')
df
```

```python
#Completion Status
status = df.groupby('CompletionStatus').size()

plt.bar(status.index, status.values,color=['orange','pink','red'],edgecolor='blue')
plt.xlabel("Completion Status")
plt.ylabel("count")
plt.title('Count of Completion Status')
plt.show()
```

```python
#location
status = df.groupby('Location').size()

plt.bar(status.index, status.values)
plt.xlabel("Location")
plt.ylabel("count")
plt.title('Count of Completion Status')
plt.show()
```

### pie chart

```python
#Completion Status
status = df.groupby('CompletionStatus').size()
plt.pie(status,labels=status.index,hatch=['**O', 'oO', 'O.O', '.||.'],explode = (0.1, 0, 0),shadow=True,autopct='%1.1f%%')
plt.show()
```

```python
#Location
status = df.groupby('Location').size()
plt.pie(status,labels=status.index,explode = (0.1, 0.1, 0,0,0),shadow=True,autopct='%1.1f%%')
plt.show()
```
