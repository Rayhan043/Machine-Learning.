## Filtering data based on condition

```python
# Nidristo korea kaukea mention korar kjonno

not_started = df.loc[df['CompletionStatus'] == 'Not Started']

not_started
```

```python
#jara comlete korcea
complete = df.loc[df['CompletionStatus'] == 'Completed']

complete
```

```python
#DS a mark > 90 and completed


complete_ds = df.loc[(df['CompletionStatus'] == 'Completed') & (df['Data Structure Marks'] >= 90)]

complete_ds
```
