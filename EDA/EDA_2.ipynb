```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv('titanic_data_updated.csv')
df.head()
```

```python
sns.countplot(x = df['Sex'], hue=df['Survived'])
```

```python
#split korar jonno groupby()

df.groupby('Sex')['Survived'].value_counts(normalize=True) #parcjentage korar jonno normalize
```

```python
df.groupby('Pclass')['Survived'].value_counts(normalize=True) #parcjentage korar jonno normalize
```

```python
sns.countplot(x= df['Pclass'], hue=df['Survived'])
```

```python
#bar plot

sns.barplot(x=df['Survived'], y=df['Age'],errorbar=None)
plt.show()
```

```python
sns.barplot(x=df['Pclass'], y=df['Age'],hue=df['Sex'],errorbar=None)
```

```python
# Density Graph
# some graph
sns.kdeplot(data =df, x=df['Age'], hue=df['Survived'])
plt.show()
```
