# Univariate Analysis(catagorical)

```python
#Pie Chart of Pclass

survival_count = df['Survived'].value_counts()
survival_label = df['Survived'].unique()
print(survival_label)
plt.pie(survival_count,labels=survival_label, autopct='%1.1f%%',shadow=True)
plt.show()
```
